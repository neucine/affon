# Tensor Selection

Selection, masking, encoding, and ranking operations.

```typescript
import { tensor, add, sub, mul, div, exp, log, sqrt, abs, neg, clamp, relu, gelu, sigmoid, tanh, sum, mean, variance, std, argmax, reshape, transpose, permute, squeeze, unsqueeze, cat, stack, contiguous, where, masked_fill, softmax, one_hot, gather, topk } from 'affon:compute/legacy'



function printValue(label: string, value: { toString(): string }) {
  console.log(label)
  console.log(value)
}

function printBeforeAfter(beforeLabel: string, beforeValue: { toString(): string }, afterLabel: string, afterValue: { toString(): string }) {
  printValue(beforeLabel, beforeValue)
  printValue(afterLabel, afterValue)
}
```

## where

Select from two tensors using a condition tensor.

```typescript
const cond = tensor([[[1, 0], [0, 1]], [[0, 1], [1, 0]]])
const a = tensor([[[10, 20], [30, 40]], [[50, 60], [70, 80]]])
const b = tensor([[[100, 200], [300, 400]], [[500, 600], [700, 800]]])
printBeforeAfter('before condition', cond, 'after where(cond, a, b)', where(cond, a, b))
```

## masked_fill

Replace masked positions with a fixed value.

```typescript
const input = tensor([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
const mask = tensor([[[0, 1], [1, 0]], [[1, 0], [0, 1]]])
printBeforeAfter('before input', input, 'after masked_fill(input, mask, -1)', masked_fill(input, mask, -1))
```

## one_hot

Append a class dimension from index values.

```typescript
const index = tensor([[0, 2], [1, 0]])
printBeforeAfter('before index', index, 'after one_hot(index, 3)', one_hot(index, 3))
```

## gather

Select values along one axis using an index tensor.

```typescript
const input = tensor([[[10, 20], [30, 40]], [[50, 60], [70, 80]]])
const index = tensor([[[1, 0], [0, 1]], [[0, 1], [1, 0]]])
printBeforeAfter('before input', input, 'after gather(input, 2, index)', gather(input, 2, index))
```

## topk

Return the largest `k` values and indices along one axis.

```typescript
const input = tensor([[[10, 30, 20], [60, 40, 50]], [[15, 25, 5], [8, 12, 10]]])
const result = topk(input, 2, 2)
printValue('before input', input)
printValue('after topk values', result.values)
printValue('after topk indices', result.indices)
```
