# Tensor Activations

Activation examples with printed values and plotted curves.


```typescript
import { tensor, relu, sigmoid, tanh, gelu, linspace } from 'affon:compute'
import { figure, title, xlabel, ylabel, plot, show } from 'std:plot'



function printBeforeAfter(beforeLabel: string, beforeValue: { toString(): string }, afterLabel: string, afterValue: { toString(): string }) {
  console.log(beforeLabel)
  console.log(beforeValue.toString())
  console.log(afterLabel)
  console.log(afterValue.toString())
}

function plotActivation(name: string, fn: (x: ReturnType<typeof tensor>) => ReturnType<typeof tensor>) {
  const xs = linspace(-4, 4, 200)
  const ys = fn(xs)
  figure({ width: 640, height: 360 })
  plot(xs, ys, { label: name, color: '#1f77b4' })
  title(name)
  xlabel('x')
  ylabel('y')
  return show()
}
```

## ReLU

$$
\mathrm{ReLU}(x) = \max(x, 0)
$$


```typescript
const reluInput = tensor([-2, -1, 0, 1, 2])
const reluOutput = relu(reluInput)
printBeforeAfter('before input', reluInput, 'after relu', reluOutput)
plotActivation('ReLU', relu)
```

## Sigmoid

$$
\sigma(x) = \frac{1}{1 + e^{-x}}
$$


```typescript
const sigmoidInput = tensor([-4, -1, 0, 1, 4])
const sigmoidOutput = sigmoid(sigmoidInput)
printBeforeAfter('before input', sigmoidInput, 'after sigmoid', sigmoidOutput)
plotActivation('Sigmoid', sigmoid)
```

## Tanh

$$
\tanh(x) = \frac{e^{2x} - 1}{e^{2x} + 1}
$$


```typescript
const tanhInput = tensor([-4, -1, 0, 1, 4])
const tanhOutput = tanh(tanhInput)
printBeforeAfter('before input', tanhInput, 'after tanh', tanhOutput)
plotActivation('Tanh', tanh)
```

## GELU

Plain text:

`0.5 * x * (1 + tanh(sqrt(2 / pi) * (x + 0.044715 * x^3)))`

$$
\mathrm{GELU}(x) = \frac{1}{2}x\left(1+\tanh\left(\sqrt{\frac{2}{\pi}}(x+0.044715x^3)\right)\right)
$$


```typescript
const geluInput = tensor([-4, -1, 0, 1, 4])
const geluOutput = gelu(geluInput)
printBeforeAfter('before input', geluInput, 'after gelu', geluOutput)
plotActivation('GELU', gelu)
```
