# Tensor Reduction

Reduction and normalization examples on 3D tensors.

```typescript
import { tensor, add, sub, mul, div, exp, log, sqrt, abs, neg, clamp, relu, gelu, sigmoid, tanh, sum, mean, variance, std, argmax, reshape, transpose, permute, squeeze, unsqueeze, cat, stack, contiguous, where, masked_fill, softmax, one_hot, gather, topk } from 'affon:compute'



function printValue(label: string, value: { toString(): string }) {
  console.log(label)
  console.log(value)
}

function printBeforeAfter(beforeLabel: string, beforeValue: { toString(): string }, afterLabel: string, afterValue: { toString(): string }) {
  printValue(beforeLabel, beforeValue)
  printValue(afterLabel, afterValue)
}
```

## sum

$$
\sum x
$$


```typescript
const x = tensor([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
printBeforeAfter('before x', x, 'after sum(x, axis=2)', sum(x, 2))
```

## mean

$$
\mathrm{mean}(x)
$$


```typescript
const x = tensor([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
printBeforeAfter('before x', x, 'after mean(x, axis=1)', mean(x, 1))
```

## variance

$$
\mathrm{Var}(x) = \mathrm{mean}((x - \mathrm{mean}(x))^2)
$$


```typescript
const x = tensor([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
printBeforeAfter('before x', x, 'after variance(x, axis=2)', variance(x, 2))
```

## std

$$
\mathrm{std}(x) = \sqrt{\mathrm{Var}(x)}
$$


```typescript
const x = tensor([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
printBeforeAfter('before x', x, 'after std(x, axis=2)', std(x, 2))
```

## argmax

Return the index of the maximum value along an axis.

```typescript
const x = tensor([[[3, 2], [8, 1]], [[5, 9], [7, 4]]])
printBeforeAfter('before x', x, 'after argmax(x, axis=2)', argmax(x, 2))
```

## softmax

$$
\mathrm{softmax}(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}
$$


```typescript
const logits = tensor([[[1, 2, 3], [2, 1, 0]], [[0, 1, 2], [3, 2, 1]]])
printBeforeAfter('before logits', logits, 'after softmax(logits, 2)', softmax(logits, 2))
```
