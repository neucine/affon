# Tensor Gradients

Small parameter-gradient examples for `affon:compute`.


```typescript
import { parameter, add, matmul, sum, mean, grad } from 'affon:compute'
```

## Add Backward

$$
z = x + y,
\quad L = \sum z
$$


```typescript
const x = parameter([2, 2, 2]).randn()
const y = parameter([2, 2, 2]).randn()
const z = add(x, y)
const loss = sum(z)
console.log('before loss')
console.log(loss)

grad(loss, [x, y])
console.log('dx =', x.grad)
console.log('dy =', y.grad)
```

## Matmul Backward

$$
Y = XW,
\quad L = \mathrm{mean}(Y)
$$


```typescript
const x = parameter([2, 2, 3]).randn()
const w = parameter([2, 3, 2]).randn()
const y = matmul(x, w)
const loss = mean(y)
console.log('before loss')
console.log(loss)

grad(loss, [x, w])
console.log('dx =', x.grad)
console.log('dw =', w.grad)
```
