# Tensor Losses

Loss examples using `affon:nn/legacy` factories.


```typescript
import { tensor } from 'affon:compute/legacy'
import nn from 'affon:nn/legacy'

```

## MSE Loss

$$
\mathrm{MSE} = \frac{1}{N}\sum_i (\hat{y}_i - y_i)^2
$$


```typescript
const criterion = nn.MSELoss()
const pred = tensor([[[1.0], [2.0]], [[3.0], [4.0]]])
const target = tensor([[[1.5], [1.5]], [[2.5], [3.5]]])
console.log('before prediction')
console.log(pred)
console.log('before target')
console.log(target)
console.log('after MSE loss')
console.log(criterion(pred, target))
```

## BCE Loss

$$
\mathrm{BCE} = -\frac{1}{N}\sum_i \left[y_i \log p_i + (1-y_i)\log(1-p_i)\right]
$$


```typescript
const criterion = nn.BCELoss()
const pred = tensor([[[0.9], [0.2]], [[0.7], [0.4]]])
const target = tensor([[[1.0], [0.0]], [[1.0], [0.0]]])
console.log('before probabilities')
console.log(pred)
console.log('before target')
console.log(target)
console.log('after BCE loss')
console.log(criterion(pred, target))
```

## BCE With Logits Loss

Plain text:

`mean(max(x, 0) - x * y + log(1 + exp(-abs(x))))`

$$
\frac{1}{N}\sum_i \left(\max(x_i,0) - x_i y_i + \log(1 + e^{-|x_i|})\right)
$$


```typescript
const criterion = nn.BCEWithLogitsLoss()
const logits = tensor([[[2.0], [-1.5]], [[0.8], [-0.3]]])
const target = tensor([[[1.0], [0.0]], [[1.0], [0.0]]])
console.log('before logits')
console.log(logits)
console.log('before target')
console.log(target)
console.log('after BCEWithLogits loss')
console.log(criterion(logits, target))
```

## Cross Entropy Loss

This example uses indexed class targets with an explicit 
`target: 'index'` option.

$$
\mathrm{CE} = -\frac{1}{N}\sum_i \log \mathrm{softmax}(x_i)_{y_i}
$$


```typescript
const criterion = nn.CrossEntropyLoss({ target: 'index' })
const logits = tensor([
  [2.0, 0.5, -1.0],
  [0.1, 1.2, 0.3],
  [1.5, 0.2, -0.4],
  [0.3, 0.7, 1.8],
])
const target = tensor([0, 1, 0, 2], 'i64')
console.log('before logits')
console.log(logits)
console.log('before indexed targets')
console.log(target)
console.log('after cross entropy loss')
console.log(criterion(logits, target))
```
