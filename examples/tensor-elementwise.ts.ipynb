# Tensor Elementwise

Elementwise arithmetic, unary math, and activation ops.

```typescript
import { tensor, add, sub, mul, div, exp, log, sqrt, abs, neg, clamp, relu, gelu, sigmoid, tanh, sum, mean, variance, std, argmax, reshape, transpose, permute, squeeze, unsqueeze, cat, stack, contiguous, where, masked_fill, softmax, one_hot, gather, topk } from 'affon:compute/legacy'



function printValue(label: string, value: { toString(): string }) {
  console.log(label)
  console.log(value)
}

function printBeforeAfter(beforeLabel: string, beforeValue: { toString(): string }, afterLabel: string, afterValue: { toString(): string }) {
  printValue(beforeLabel, beforeValue)
  printValue(afterLabel, afterValue)
}
```

## add

$$
z = x + y
$$


```typescript
const x = tensor([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
const y = tensor([[[10, 20], [30, 40]], [[50, 60], [70, 80]]])
printBeforeAfter('before x', x, 'after add(x, y)', add(x, y))
```

## gelu

$$

$$


```typescript
const x = tensor([[[-2, -1], [0, 1]], [[2, 3], [-4, 5]]])
printBeforeAfter('before x', x, 'after gelu(x)', gelu(x))
```

## tanh

$$

```typescript
const x = tensor([[[-4, -1], [0, 1]], [[2, 3], [-2, 4]]])
printBeforeAfter('before x', x, 'after tanh(x)', tanh(x))
```
